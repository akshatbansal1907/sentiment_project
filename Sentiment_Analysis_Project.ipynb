```python
import torch
import transformers
import gradio
import fastapi

print("All libraries are working!")
```

```python
from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

print("Sentiment model loaded successfully!")
```

```python
def predict_sentiment(text):
    result = classifier(text)[0]

    return {
        "label": result["label"],
        "confidence": float(result["score"])
    }
```

```python
result = predict_sentiment("I really love this product!")

print(result)
```

```python
result = predict_sentiment("This product is terrible and I hate it.")

print(result)
```

```python
import gradio as gr

def predict(text):
    result = predict_sentiment(text)

    return (
        f"Sentiment: {result['label']}\n"
        f"Confidence: {result['confidence']:.3f}"
    )

demo = gr.Interface(
    fn=predict,
    inputs=gr.Textbox(
        lines=5,
        placeholder="Enter a sentence or review..."
    ),
    outputs=gr.Textbox(),
    title="Sentiment Analysis",
    description="Enter text to determine whether the sentiment is positive or negative."
)

demo.launch()
```

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(
    title="Sentiment Analysis API",
    description="API for sentiment analysis using DistilBERT",
    version="1.0.0"
)


class TextInput(BaseModel):
    text: str


@app.get("/")
def home():
    return {
        "message": "Sentiment Analysis API is running"
    }


@app.post("/predict")
def predict_api(data: TextInput):
    return predict_sentiment(data.text)
```

```python
import threading
import uvicorn

def run_api():
    uvicorn.run(app, host="127.0.0.1", port=8000)

api_thread = threading.Thread(target=run_api, daemon=True)
api_thread.start()

print("FastAPI server started!")
```

```python
%%writefile model.py

from transformers import pipeline

classifier = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)


def predict_sentiment(text):
    result = classifier(text)[0]

    return {
        "label": result["label"],
        "confidence": float(result["score"])
    }
```
